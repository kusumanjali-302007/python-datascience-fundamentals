```python
a = 10
b = 3

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus (remainder):", a % b)
print("Power:", a ** b)


print(f"{a} divided by {b} is {a/b:.2f}")
```
