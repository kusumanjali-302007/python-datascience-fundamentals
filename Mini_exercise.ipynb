```python
import csv

# Create a sample CSV first (skip if you already have one)
with open("students.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["name", "score"])
    writer.writerow(["Alex", 85])
    writer.writerow(["Priya", 92])
    writer.writerow(["Sam", 78])

# Read and summarize it
total = 0
count = 0

with open("students.csv", "r") as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(f"{row['name']} scored {row['score']}")
        total += int(row['score'])
        count += 1

print(f"\nAverage score: {total / count:.2f}")
```
